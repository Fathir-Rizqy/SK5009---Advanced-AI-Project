```python
import numpy as np
import matplotlib.pyplot as plt


```

```python
koefisien = [1, 5, 2, 4, 2]
p = np.poly1d(koefisien)
print("Polinom")
print(f"{p}")
```

```python
p_turunan1 = p.deriv()
p_turunan2 = p_turunan1.deriv()

print(f"Turunan Pertama (f'(x)):")
print(f"{p_turunan1}")
print(f"Turunan Kedua (f''(x)):")
print(f"{p_turunan2}")


titik_kritis = np.roots(p_turunan1)
print("Titik kritis")
print(f"{titik_kritis}")

```

```python
maxima_lokal = []
minima_lokal = []
for x in titik_kritis:
    nilai_turunan2 = round(p_turunan2(x), 2)
    if nilai_turunan2 < 0:
        y = p(x)
        print(f"Uji di x = {x:.2f}: f''({x:.2f}) = {nilai_turunan2} (< 0). Ini adalah MAKSIMUM lokal.")
        print(f"   Titik maksimum lokal: ({x:.2f}, {y:.2f})\n")
        maxima_lokal.append((x, y))
    elif nilai_turunan2 > 0:
        y = p(x)
        print(f"Uji di x = {x:.2f}: f''({x:.2f}) = {nilai_turunan2} (> 0). Ini adalah MINIMUM lokal.")
        print(f"   Titik minimum lokal: ({x:.2f}, {y:.2f})\n")
        minima_lokal.append((x, y))
    else:
        print(f"Uji di x = {x:.2f}: f''({x:.2f}) = {nilai_turunan2}. Uji turunan kedua gagal.")

```

```python
x_vals = np.linspace(min(titik_kritis) - 2, max(titik_kritis) + 2, 400)
y_vals = p(x_vals)

fig, ax = plt.subplots()

ax.plot(x_vals, y_vals)

if minima_lokal:
    min_x, min_y = zip(*minima_lokal)
    ax.scatter(min_x, min_y, color='red', s=100, zorder=5, label='Minimum Lokal')

if maxima_lokal:
    max_x, max_y = zip(*maxima_lokal)
    ax.scatter(max_x, max_y, color='green', s=100, zorder=5, label='Maksimum Lokal')

ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.legend()
ax.grid(True)

plt.show()
```
